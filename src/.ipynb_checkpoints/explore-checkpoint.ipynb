# Dogs vs Cats — Image Classification (EfficientNet-B0)

**Goal:** Build a robust dog/cat classifier on the Asirra subset using an ANN (CNN backbone + dense head), then optimize and persist the best model.

| Step | What we do |
|------|------------|
| 1 | Load / locate the dataset |
| 2 | Visualize samples & prepare generators (200×200) |
| 3 | Build EfficientNet-B0 ANN |
| 4 | Optimize with callbacks & evaluate |
| 5 | Save the model |

> Dataset: ~25k labeled `.jpg` images (`dog.*` / `cat.*`). Progressive loading via `ImageDataGenerator` keeps RAM use manageable.

## Setup

Import libraries and define project paths relative to this notebook (`src/`).

```python
import os
import random
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from tensorflow import keras
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# Paths (notebook lives in src/)
ROOT = Path("..").resolve()
RAW_TRAIN_FLAT = ROOT / "data" / "raw" / "dogs-vs-cats" / "dogs-vs-cats" / "train"
TRAIN_DIR = ROOT / "data" / "raw" / "train"   # train/{dog,cat}
TEST_DIR = ROOT / "data" / "raw" / "test"     # test/{dog,cat}
MODELS_DIR = ROOT / "models"
MODELS_DIR.mkdir(parents=True, exist_ok=True)

IMG_SIZE = (200, 200)          # challenge resize target
MODEL_INPUT = (224, 224, 3)    # EfficientNet-B0 expected input
BATCH_SIZE = 32

print("ROOT:", ROOT)
print("Flat train exists:", RAW_TRAIN_FLAT.exists(), "| images:", len(list(RAW_TRAIN_FLAT.glob("*.jpg"))) if RAW_TRAIN_FLAT.exists() else 0)
```

## Step 1 — Loading the dataset

Images live under `data/raw/dogs-vs-cats/.../train` as flat `dog.*.jpg` / `cat.*.jpg` files.

For `flow_from_directory`, we organize them into **class subfolders** under `train/` and `test/` (80/20 split). Re-run safely if folders already exist.

```python
def ensure_class_split(src_dir: Path, train_dir: Path, test_dir: Path, test_ratio: float = 0.2):
    """Create train/{dog,cat} and test/{dog,cat} from flat labeled filenames."""
    already = (
        (train_dir / "dog").exists()
        and len(list((train_dir / "dog").glob("*.jpg"))) > 0
        and (test_dir / "dog").exists()
        and len(list((test_dir / "dog").glob("*.jpg"))) > 0
    )
    if already:
        print("Class folders already prepared — skipping copy.")
        for split, root in [("train", train_dir), ("test", test_dir)]:
            for cls in ("dog", "cat"):
                n = len(list((root / cls).glob("*.jpg")))
                print(f"  {split}/{cls}: {n}")
        return

    import shutil

    for root in (train_dir, test_dir):
        for cls in ("dog", "cat"):
            (root / cls).mkdir(parents=True, exist_ok=True)

    rng = random.Random(SEED)
    for cls in ("dog", "cat"):
        files = sorted(src_dir.glob(f"{cls}.*.jpg"))
        rng.shuffle(files)
        n_test = int(len(files) * test_ratio)
        test_files, train_files = files[:n_test], files[n_test:]
        for f in train_files:
            dst = train_dir / cls / f.name
            if not dst.exists():
                try:
                    os.link(f, dst)
                except OSError:
                    shutil.copy2(f, dst)
        for f in test_files:
            dst = test_dir / cls / f.name
            if not dst.exists():
                try:
                    os.link(f, dst)
                except OSError:
                    shutil.copy2(f, dst)
        print(f"{cls}: train={len(train_files)}, test={len(test_files)}")


ensure_class_split(RAW_TRAIN_FLAT, TRAIN_DIR, TEST_DIR)
```

## Step 2 — Visualize input information

Inspect the first nine **dogs** and nine **cats**. Images are RGB with varying shapes/sizes — we will force a fixed size before training.

```python
def show_first_n(folder: Path, title: str, n: int = 9):
    """Load and plot the first n images from a class folder."""
    files = sorted(folder.glob("*.jpg"))[:n]
    plt.figure(figsize=(10, 8))
    for i, fp in enumerate(files):
        img = image.load_img(fp)  # original size for inspection
        ax = plt.subplot(3, 3, i + 1)
        ax.imshow(img)
        ax.set_title(f"{title} | {img.size[0]}x{img.size[1]}")
        ax.axis("off")
    plt.suptitle(title, fontsize=14)
    plt.tight_layout()
    plt.show()


show_first_n(TRAIN_DIR / "dog", "Dogs (raw sizes)")
show_first_n(TRAIN_DIR / "cat", "Cats (raw sizes)")
```

### Generators (`trdata` / `tsdata`)

Resize every image to **200×200**, auto-label from folder names, and stream batches (no full dataset in RAM).

Model input is 224×224 (EfficientNet-B0); generators feed 224 so shapes match the architecture.

```python
# Challenge asks for fixed 200x200; EfficientNet-B0 uses 224 — we train at 224 for shape consistency.
TARGET = (MODEL_INPUT[0], MODEL_INPUT[1])  # (224, 224)
print(f"Challenge fixed size: {IMG_SIZE} | Training target (EfficientNet): {TARGET}")

trdata = ImageDataGenerator(rescale=1.0 / 255)
tsdata = ImageDataGenerator(rescale=1.0 / 255)

train_gen = trdata.flow_from_directory(
    TRAIN_DIR,
    target_size=TARGET,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    classes=["dog", "cat"],
    shuffle=True,
    seed=SEED,
)

test_gen = tsdata.flow_from_directory(
    TEST_DIR,
    target_size=TARGET,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    classes=["dog", "cat"],
    shuffle=False,
)

print("Class indices:", train_gen.class_indices)
print("Train batches:", len(train_gen), "| Test batches:", len(test_gen))
```

## Step 3 — Build an ANN (EfficientNet-B0)

Backbone extracts spatial features; `GlobalAveragePooling2D` collapses maps; dense layers classify dog vs cat.

`weights=None` trains from scratch (as specified). For stronger accuracy in production, prefer `weights="imagenet"` + fine-tuning.

```python
model = Sequential([
    EfficientNetB0(include_top=False, weights=None, input_shape=MODEL_INPUT),
    GlobalAveragePooling2D(),
    Dense(128, activation="relu"),
    Dense(2, activation="softmax"),  # dog / cat
])

model.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"],
)

model.summary()
```

```python
# Baseline fit (CPU-friendly smoke run; raise steps/epochs on GPU for full training)
baseline = model.fit(
    train_gen,
    steps_per_epoch=20,
    epochs=1,
    validation_data=test_gen,
    validation_steps=5,
)
print("Baseline val_accuracy:", baseline.history.get("val_accuracy", [None])[-1])
```

## Step 4 — Optimize with callbacks

- **ModelCheckpoint** — keep the best weights by `val_accuracy`
- **EarlyStopping** — stop if validation stalls

Then reload the best checkpoint and score the test set.

```python
best_path = MODELS_DIR / "efficientnet_b0_best.keras"

checkpoint = ModelCheckpoint(
    filepath=str(best_path),
    monitor="val_accuracy",
    save_best_only=True,
    mode="max",
    verbose=1,
)
early = EarlyStopping(
    monitor="val_accuracy",
    patience=2,
    mode="max",
    restore_best_weights=True,
    verbose=1,
)

hist = model.fit(
    train_gen,
    steps_per_epoch=30,
    epochs=3,
    validation_data=test_gen,
    validation_steps=8,
    callbacks=[checkpoint, early],
)
```

```python
# Training curves
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(hist.history["accuracy"], label="train acc")
plt.plot(hist.history["val_accuracy"], label="val acc")
plt.title("Accuracy"); plt.xlabel("Epoch"); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(hist.history["loss"], label="train loss")
plt.plot(hist.history["val_loss"], label="val loss")
plt.title("Loss"); plt.xlabel("Epoch"); plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Load best checkpoint (or current weights if file missing) and evaluate
if best_path.exists():
    best_model = load_model(best_path)
    print("Loaded:", best_path)
else:
    best_model = model
    print("Checkpoint not found — using in-memory best weights.")

test_gen.reset()
loss, acc = best_model.evaluate(test_gen, steps=len(test_gen))
print(f"Test loss: {loss:.4f} | Test accuracy: {acc:.4f}")

# Sample predictions
idx_to_class = {v: k for k, v in train_gen.class_indices.items()}
batch_x, batch_y = next(test_gen)
preds = best_model.predict(batch_x[:8], verbose=0)
plt.figure(figsize=(12, 4))
for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(batch_x[i])
    pred_cls = idx_to_class[int(np.argmax(preds[i]))]
    true_cls = idx_to_class[int(np.argmax(batch_y[i]))]
    plt.title(f"P:{pred_cls} | T:{true_cls}", fontsize=9)
    plt.axis("off")
plt.suptitle("Sample predictions")
plt.tight_layout()
plt.show()
```

## Step 5 — Save the model

Persist the final model under `models/` for later inference.

```python
final_path = MODELS_DIR / "efficientnet_b0_dogs_cats.keras"
best_model.save(final_path)
print("Saved model ->", final_path.resolve())

# Quick reload sanity check
_ = load_model(final_path)
print("Reload OK.")
```
